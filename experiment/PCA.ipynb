```python
import yaml
import torch
import sys
from pathlib import Path
sys.path.append('/home/atin-ct3/action_recognition')

from model import get_model
```

```python
# Load X3D model
config_path = "/home/atin-ct3/action_recognition/config/X3D/rfw.yaml"
checkpoint_path = "/home/atin-ct3/action_recognition/checkpoints/x3d_rfw_violence/best_model_acc.pth"

with open(config_path, 'r') as f:
    config = yaml.safe_load(f)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

print(f"Loading X3D model from {checkpoint_path}...")
model = get_model(**config)
checkpoint = torch.load(checkpoint_path, map_location=device)
model.load_state_dict(checkpoint['model_state_dict'])
model.to(device)
model.eval()

print("Model loaded successfully!")
```

```python
# Access and inspect the head architecture
head = model.model.blocks[5]
print(head)
print(f"\nHead output classes: {head.proj.out_features}")
print(f"Input features to projection: {head.proj.in_features}")
```
